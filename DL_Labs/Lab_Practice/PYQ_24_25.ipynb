```python
import numpy as np
import torch 
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as Optimizer

```

```python
#Synthetic Dataset Generation and One hot encoding

n_samples=6000
classes=6
batch_size=128
epochs=10
learning_rate=0.05

X=torch.rand(n_samples,1,32,32)

y=torch.randint(0,classes,(n_samples,))

print(X.shape)
print(y.shape)
```

```python
model=nn.Sequential(
    #Conv Layer1
    nn.Conv2d(in_channels=1,padding=1,stride=1,out_channels=16,kernel_size=3),
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2,stride=2),

    #Conv Layer2
    nn.Conv2d(in_channels=16,out_channels=32,kernel_size=3,padding=1,stride=1),
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2,stride=2),

    #flatten
    nn.Flatten(),

    nn.Linear(32*8*8,6) 
)

print(model)
```

```python
loss_func=nn.CrossEntropyLoss()

optimizer=Optimizer.SGD(model.parameters(),lr=5)

```

```python
for epoch in range(epochs):
    total_loss=0

    for i in range(0,n_samples,batch_size):
        x_curr_batch=X[i:i+batch_size]
        y_curr_batch=y[i:i+batch_size]

        #Forward
        outputs=model(x_curr_batch)
        loss=loss_func(outputs,y_curr_batch)

        #Backward
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss+=loss.item()
    

    print(f"Epoch {epoch+1}/{epochs}, Loss = {total_loss:.4f}")
```
